```python
import os
import sys
from pathlib import Path


def load_repo_environment() -> None:
    cwd = Path.cwd().resolve()
    candidates = [cwd, *cwd.parents]
    notebook_path = Path("examples/basic/hello_world_jupyter.ipynb")

    for directory in candidates:
        if not (directory / notebook_path).exists():
            continue

        src_path = directory / "src"
        if src_path.exists():
            src = str(src_path)
            if src not in sys.path:
                sys.path.insert(0, src)

        venv_path = directory / ".venv"
        site_packages = next(venv_path.glob("lib/python*/site-packages"), None)
        if site_packages is not None:
            site = str(site_packages)
            if site not in sys.path:
                sys.path.insert(1 if src_path.exists() else 0, site)

        env_path = directory / ".env"
        if env_path.exists():
            for line in env_path.read_text().splitlines():
                line = line.strip()
                if not line or line.startswith("#") or "=" not in line:
                    continue
                key, value = line.split("=", 1)
                os.environ.setdefault(key.strip(), value.strip().strip("\"'"))
        return

    raise RuntimeError("Run this notebook from inside the openai-agent-sdk-sample-python repository.")


load_repo_environment()
```
